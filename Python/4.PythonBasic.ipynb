```python
# Operaciones con String
cadena_1 = 'hadoop, hive, spark, pyspark'
cadena_2 = '7'
cadena_3 = 'siete'
```

```python
# La primera en mayuscula
cadena_1.capitalize()
```

```python
# En Minuscula
cadena_1.lower()
```

```python
# Texto en modo titulo
cadena_1.title()
```

```python
# Dividir texto
cadena_1.split()
```

```python
# contar cuantas 'a' hay
cadena_1.count('a')
```

```python
# longitud de la cadena
len(cadena_1)
```

```python
# Quitar letra o palabra
cadena_1.split('k')
```

```python
# Limpiar extremos
cadena_4 = '(Gobierno de Datos),      '
cadena_4.strip()
```

```python
cadena_4.strip('(), ')
```

```python
# Reemplazando texto
cadena_1.replace('spark', 'Gobierno de Datos')
```

```python
# Caso cadena 4, quitar los (), espacio, pasar mayusculas y al final en palabras
cadena_4.strip('(), ').upper().split()
```

3. Loops

```python
palabras = cadena_1.split()
palabras
```

```python
for palabra in palabras:
    print(palabra)
```

```python
# Colacar numeracion a cada palabra
```

```python
for idx, palabra in enumerate(palabras):
    print(idx, palabra)
```
